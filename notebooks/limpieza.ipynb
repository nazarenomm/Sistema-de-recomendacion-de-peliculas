Esto suele pasar por exportación con encoding html

```python
for col in df_pelis.columns:
    if df_pelis[col].dtype == 'object':  # Solo columnas string
        df_pelis[col] = df_pelis[col].apply(lambda x: html.unescape(str(x)) if pd.notna(x) else x)
```

## Peliculas faltantes

En el dataset de usuarios se mencionan películas que no tenemos en la base de datos, eso no debería ser posible en un sistema cerrado como el que planteamos

```python
df_usuarios = pd.read_csv('../data/usuarios.csv')
```

```python
# chequear si las peliculas en pelicula_1, pelicula_2,...pelicula_5 existen en el dataset de peliculas
peliculas_set = set(df_pelis['name'])
for i in range(1, 6):
    for pelicula in df_usuarios[f'pelicula_{i}']:
        if pelicula not in peliculas_set:
            print(f'La película "{pelicula}" no existe en el dataset de películas.')
```
